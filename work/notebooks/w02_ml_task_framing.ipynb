## My lane as an ML task

### Task Type
**Classification**

### Why?

My lane focuses on deciding whether a piece of content should be refreshed based on its historical performance. This is naturally a binary classification problem because every article can be classified into one of two categories:

- Needs Refresh
- Does Not Need Refresh

The model will learn patterns from previous content performance and predict whether an article is likely to benefit from being updated.

## Target / Proxy
The dataset does not include a true label indicating whether an article needs a refresh.

For this notebook, I create a simple proxy target for framing the ML problem. Articles with a downward trend greater than 30% are labeled as needing refresh.

This is only an example proxy used for problem framing, not a ground-truth label.

```python
import pandas as pd
```

```python
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

df.head()
```

```python
df.shape
```

```python
list(df.columns)
```

```python
df.info()
```

```python
df["needs_refresh"] = (
    (df["trend_direction"] == "down") &
    (df["trend_pct"] < -30)
)

df[["trend_direction", "trend_pct", "needs_refresh"]].head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*


I would evaluate this classification task using the **F1-score**.

Why?

The number of articles that actually need refreshing is expected to be much smaller than the number of healthy articles.

F1-score balances Precision and Recall, making it a better choice than Accuracy for an imbalanced dataset.

## Why Machine Learning?

A fixed rule such as:

"If an article is older than 180 days, refresh it."

would ignore many important signals.

For example:

- CTR
- Search impressions
- Clicks
- Search position
- Traffic trend
- Engagement
- Seasonality

Some old articles still perform very well, while some newer articles lose traffic quickly.

Machine Learning is appropriate because article performance depends on multiple interacting features, such as traffic, CTR, search position, trend, and content age. A single fixed rule cannot capture these relationships effectively, while an ML model can learn patterns from historical data to support better refresh decisions.

## What action does this prediction support?

The prediction helps the content team prioritize which articles should be refreshed first.

Instead of reviewing thousands of articles manually, the model can recommend the articles most likely to benefit from updating.

This saves time and helps improve SEO performance.
